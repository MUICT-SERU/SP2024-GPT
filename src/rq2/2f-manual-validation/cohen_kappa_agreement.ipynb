```python
import pandas as pd
from sklearn.metrics import cohen_kappa_score
```

```python
result_df = pd.read_csv('../dataset/sampled_354_hn_gh_ai_comment_2hu_labeled.csv')
labels = [-1, 0, 1]
score1 = cohen_kappa_score(result_df['comment_sentiment_hu2'], result_df['comment_sentiment_hu1'], labels=labels)
score2 = cohen_kappa_score(result_df['comment_sentiment_hu2'], result_df['comment_sentiment'], labels=labels)
score3 = cohen_kappa_score(result_df['comment_sentiment_hu1'], result_df['comment_sentiment'], labels=labels)
score4 = cohen_kappa_score(result_df['comment_sentiment_consensus'], result_df['comment_sentiment'], labels=labels)

print("Cohen kappa score between hu2 & hu1: " + str(score1))
print("Cohen kappa score between human consensus & gpt: " + str(score4))
```

```python
result_df = pd.read_csv('../dataset/sampled_328_hu_gh_ai_story_2hu_labeled.csv')
labels = ['Negative', 'Neutral', 'Positive']
score1 = cohen_kappa_score(result_df['story_sentiment_hu2'], result_df['story_sentiment_hu1'], labels=labels)
score2 = cohen_kappa_score(result_df['story_sentiment_hu2'], result_df['llm_sentiment'], labels=labels)
score3 = cohen_kappa_score(result_df['story_sentiment_hu1'], result_df['llm_sentiment'], labels=labels)
score4 = cohen_kappa_score(result_df['hu_consensus'], result_df['llm_sentiment'], labels=labels)

print("Cohen kappa score between hu2 & hu1: " + str(score1))
print("Cohen kappa score between human consensus & gpt: " + str(score4))
```
