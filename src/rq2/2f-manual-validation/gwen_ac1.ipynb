```python
import pandas as pd
from irrCAC.raw import CAC
```

```python
result_df = pd.read_csv('../dataset/sampled_354_hn_gh_ai_comment_2hu_labeled.csv')
df = result_df[['comment_sentiment', 'comment_sentiment_consensus']]
cac_analysis = CAC(df)
gwet_results = cac_analysis.gwet()
coefficient = gwet_results['est']['coefficient_value']

print(f"Gwet's AC1 Coefficient: {coefficient:.4f}")
```
